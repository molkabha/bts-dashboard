```python
# Cellule 1b — VERSION CORRIGÉE avec validation du signe

print('Calibration des paramètres de dynamique QoS...')

df_sorted_qos = df_2024.sort_values(['station_id', 'timestamp']).copy()
df_sorted_qos['qos_delta'] = df_sorted_qos.groupby('station_id')['score_qos'].diff()

# Delta moyen hors anomalie
delta_normal_raw = df_sorted_qos.loc[
    df_sorted_qos['label_ensemble_score'] == 0, 'qos_delta'].dropna()

# Delta moyen lors d'une anomalie
delta_anomalie_raw = df_sorted_qos.loc[
    df_sorted_qos['label_ensemble_score'] == 1, 'qos_delta'].dropna()

delta_normal_p50   = delta_normal_raw.median()
delta_anomalie_p50 = delta_anomalie_raw.median()

print(f'  Δ QoS médiane (normal)   : {delta_normal_p50:+.5f}')
print(f'  Δ QoS médiane (anomalie) : {delta_anomalie_p50:+.5f}')

# - En état normal, le delta doit être ≥ 0 (récupération ou stabilité)
# - En état anomalie, le delta doit être ≤ 0 (dégradation)
if delta_normal_p50 < 0:
    print(f'  AVERTISSEMENT : delta_normal_p50 est négatif ({delta_normal_p50:+.5f})')
    print(f'  → Cela indique que la QoS se dégrade même hors anomalie sur ce dataset.')
    print(f'  → Paramètre forcé à la valeur plancher (0.001) mais à investiguer.')

if delta_anomalie_p50 > 0:
    print(f'  AVERTISSEMENT : delta_anomalie_p50 est positif ({delta_anomalie_p50:+.5f})')
    print(f'  → Cela indique une amélioration de QoS lors d\'anomalies (contre-intuitif).')
    print(f'  → Vérifier la définition de label_ensemble_score.')

# Paramètres calibrés avec vérification du signe
QOS_DELTA_NORMAL   = max(0.001, float(delta_normal_p50))    # doit être positif (récupération)
QOS_DELTA_ANOMALIE = max(0.005, -float(delta_anomalie_p50)) # doit être positif (dégradation)

print(f'\n  → BTSEnvironnement.step() utilisera :')
print(f'      qos_next (action valide)  = score_qos + {QOS_DELTA_NORMAL:.4f}')
print(f'      qos_next (violation QoS)  = score_qos - {QOS_DELTA_ANOMALIE:.4f}')
print(f'\n  Signe delta_normal OK   : {delta_normal_p50 >= 0}')
print(f'  Signe delta_anomalie OK : {delta_anomalie_p50 <= 0}')
```

```python
# Cellule 2 — Moteur de décision
# Garde-fou QoS : le mode ECO n'est jamais attribué si optimisation_qos_autorisee == 0

class MoteurDecisionEnergie:
    SEUILS = {
        'eco_score'       : 0.25,
        'critique_score'  : 0.60,
        'eco_cpu'         : 50,
        'critique_ecart'  : 30,
        'eco_heure_debut' : 0,
        'eco_heure_fin'   : 6,
    }
    MODES = {
        'ECO'      : {'couleur': '#2ecc71', 'priorite': 0},
        'NORMAL'   : {'couleur': '#3498db', 'priorite': 1},
        'ATTENTION': {'couleur': '#f39c12', 'priorite': 2},
        'CRITIQUE' : {'couleur': '#e74c3c', 'priorite': 3},
    }
    ACTIONS = {
        'ECO': [
            'Activer Sleep Mode sur secteurs non sollicités',
            'Réduire puissance émission de 2 dBm',
            'Activer free cooling si conditions météo favorables',
            'Consolider trafic sur secteurs actifs',
        ],
        'NORMAL'   : ['Monitoring standard toutes les 15 min', 'Maintenir configuration nominale'],
        'ATTENTION': ['Augmenter fréquence monitoring (5 min)', 'Préparer plan de basculement', 'Alerter superviseur'],
        'CRITIQUE' : ['ALERTE IMMEDIATE — Intervention requise', 'Notifier NOC', 'Activer redondance', 'Ticket maintenance prioritaire'],
    }

    def __init__(self, qos_seuil: float = 0.60):
        self.qos_seuil = qos_seuil

    def _calculer_risque_qos(self, score_qos: float) -> str:
        if score_qos >= 0.80:            return 'Faible'
        elif score_qos >= self.qos_seuil: return 'Modéré'
        else:                             return 'Critique'

    def decider_mode(self, row: pd.Series) -> dict:
        score     = row.get('anomalie_score_ensemble', 0)
        ecart     = abs(row.get('ecart_pct', 0))
        cpu       = row.get('charge_cpu_pct', 50)
        heure     = row.get('heure', 12)
        votes     = row.get('nb_votes_anomalie', 0)
        score_qos = float(row.get('score_qos', 1.0))
        qos_ok    = int(row.get('optimisation_qos_autorisee', 1))
        risque_qos = self._calculer_risque_qos(score_qos)

        if score > self.SEUILS['critique_score'] or votes >= 5:
            mode = 'CRITIQUE'
        elif ecart > self.SEUILS['critique_ecart'] or (score > self.SEUILS['eco_score'] and votes >= 2):
            mode = 'ATTENTION'
        elif (score < self.SEUILS['eco_score'] and
              cpu < self.SEUILS['eco_cpu'] and
              self.SEUILS['eco_heure_debut'] <= heure <= self.SEUILS['eco_heure_fin'] and
              qos_ok == 1):
            mode = 'ECO'
        else:
            mode = 'NORMAL'

        eco_pct = {'ECO': 20, 'NORMAL': 0, 'ATTENTION': 0, 'CRITIQUE': 0}[mode]
        return {
            'mode'             : mode,
            'priorite'         : self.MODES[mode]['priorite'],
            'action_principale': self.ACTIONS[mode][0],
            'actions_full'     : self.ACTIONS[mode],
            'score_anomalie'   : round(score, 4),
            'eco_potentiel_pct': eco_pct,
            'score_qos'        : round(score_qos, 4),
            'risque_qos'       : risque_qos,
        }

    def appliquer_sur_dataset(self, df_in: pd.DataFrame) -> pd.DataFrame:
        df_out = df_in.copy()
 
        # Extraction vectorisée des colonnes nécessaires
        score  = df_out.get('anomalie_score_ensemble', pd.Series(0.0, index=df_out.index))
        ecart  = df_out.get('ecart_pct', pd.Series(0.0, index=df_out.index)).abs()
        cpu    = df_out.get('charge_cpu_pct', pd.Series(50.0, index=df_out.index))
        heure  = df_out.get('heure', pd.Series(12, index=df_out.index))
        votes  = df_out.get('nb_votes_anomalie', pd.Series(0, index=df_out.index))
        sq     = df_out.get('score_qos', pd.Series(1.0, index=df_out.index)).astype(float)
        qos_ok = df_out.get('optimisation_qos_autorisee', pd.Series(1, index=df_out.index)).astype(int)
 
        S = self.SEUILS
        mask_critique  = (score > S['critique_score']) | (votes >= 5)
        mask_attention = (~mask_critique) & (
            (ecart > S['critique_ecart']) | ((score > S['eco_score']) & (votes >= 2))
        )
        mask_eco = (
            (~mask_critique) & (~mask_attention) &
            (score < S['eco_score']) &
            (cpu < S['eco_cpu']) &
            (heure >= S['eco_heure_debut']) & (heure <= S['eco_heure_fin']) &
            (qos_ok == 1)
        )
 
        df_out['mode_operation'] = np.select(
            [mask_critique, mask_attention, mask_eco],
            ['CRITIQUE',    'ATTENTION',    'ECO'],
            default='NORMAL'
            )
        df_out['priorite'] = df_out['mode_operation'].map(
            {m: v['priorite'] for m, v in self.MODES.items()}
        )
        df_out['action_principale'] = df_out['mode_operation'].map(
            {m: v[0] for m, v in self.ACTIONS.items()}
        )
        df_out['eco_potentiel_pct'] = df_out['mode_operation'].map(
            {'ECO': 20, 'NORMAL': 0, 'ATTENTION': 0, 'CRITIQUE': 0}
        )
        df_out['risque_qos'] = pd.cut(
            sq,
            bins=[-np.inf, self.qos_seuil, 0.80, np.inf],
            labels=['Critique', 'Modéré', 'Faible']
        ).astype(str)
        return df_out


moteur = MoteurDecisionEnergie(qos_seuil=qos_seuil)

print(f'Moteur instancié avec qos_seuil={moteur.qos_seuil}')
print('Application du moteur de décision sur le jeu 2024...')
t0 = time.time()
df_2024 = moteur.appliquer_sur_dataset(df_2024)
print(f'Terminé en {time.time()-t0:.1f}s')
print()
print('Distribution des modes opérationnels :')
for mode, n in df_2024['mode_operation'].value_counts().items():
    print(f'  {mode:10s} : {n:>8,} ({n/len(df_2024)*100:5.1f}%)')
print()
print('Distribution risque QoS :')
print(df_2024['risque_qos'].value_counts().to_string())
```

```python
# Cellule 3 — 5 stratégies d'optimisation calibrées sur le dataset réel

class StrategieOptimisation:

    # Économie kWh par heure de sleep-mode d'un secteur actif
    # Source : GSMA Energy Efficiency Report (2022), Table 4-3 ;
    #          ajustements terrain estimés sur df_2024 (consommation moyenne
    #          par technologie × fraction idle mesurée hors trafic 01h–05h).
    #   2G  : ~0.06 kWh/h (faible puissance, BTS legacy)
    #   3G  : ~1.6  kWh/h (NodeB actif : ~3.2 kW × 50 % idle)
    #   4G  : ~1.8  kWh/h (eNodeB : ~3.7 kW × 50 % idle)
    #   4G+ : ~2.0  kWh/h (eNodeB CA : ~4.1 kW × 50 % idle)
    ECO_SLEEP_PAR_TECHNO = {'2G': 0.062, '3G': 1.608, '4G': 1.845, '4G+': 2.040}
    _ORDRE_TECHNO        = ['2G', '3G', '4G', '4G+']

    # ── NOTE : les assert de validation sont placés APRÈS la classe ──
    # (dans un corps de classe, les générateurs créent leur propre portée
    #  et ne voient pas les noms de classe → NameError inévitable si inline)
    def __init__(self, qos_seuil: float = None):
        self.qos_seuil = qos_seuil if qos_seuil is not None else globals().get('qos_seuil', 0.60)

    def _get_col(self, df, col, default=None):
        if col in df.columns: return df[col]
        if default is not None: return pd.Series(default, index=df.index)
        raise KeyError(f'Colonne manquante : {col}')

    def appliquer_strategies(self, df_in: pd.DataFrame) -> pd.DataFrame:
        df_out = df_in.copy()
        qos_ok = self._get_col(df_out, 'optimisation_qos_autorisee', default=1)

        df_out['action_proposee']       = 'aucune_action'
        df_out['economie_estimee_kwh']  = 0.0
        df_out['puissance_cible_dbm']   = df_out.get('puissance_emission_dbm', np.nan)
        df_out['priorite_intervention'] = ''

        # STRATÉGIE 1 — Sleep mode secteur
        peut_sleep = (
            (qos_ok == 1) &
            (df_out['heure'].between(1, 5)) &
            (self._get_col(df_out, 'taux_charge_voix', 0) < 0.15) &
            (self._get_col(df_out, 'taux_charge_data', 0) < 0.15) &
            (self._get_col(df_out, 'score_qos', 1)         > 0.80) &
            (self._get_col(df_out, 'nb_secteurs_actifs', 2) > 1)
        )
        eco_sleep = df_out.loc[peut_sleep, 'technologie'].map(self.ECO_SLEEP_PAR_TECHNO).fillna(0)
        df_out.loc[peut_sleep, 'action_proposee']      = 'sleep_mode_secteur'
        df_out.loc[peut_sleep, 'economie_estimee_kwh'] = eco_sleep
        self.couverture_1 = peut_sleep.mean()

        # STRATÉGIE 2 — Réduction de puissance d'émission
        peut_reduire_puissance = (
            (qos_ok == 1) &
            (df_out['technologie'].isin(['4G', '4G+'])) &
            (self._get_col(df_out, 'puissance_emission_dbm', 43) > 40.5) &
            (self._get_col(df_out, 'score_qos', 1)  > 0.75) &
            (self._get_col(df_out, 'charge_cpu_pct', 0) < 50.0) &
            (self._get_col(df_out, 'taux_charge_data', 0) < 0.60) &
            (~peut_sleep)
        )
        df_out.loc[peut_reduire_puissance, 'action_proposee']      = 'reduction_puissance'
        df_out.loc[peut_reduire_puissance, 'economie_estimee_kwh'] = (
            df_out.loc[peut_reduire_puissance, 'consommation_kwh'] * 0.08
        )
        if 'puissance_emission_dbm' in df_out.columns:
            df_out.loc[peut_reduire_puissance, 'puissance_cible_dbm'] = (
                df_out.loc[peut_reduire_puissance, 'puissance_emission_dbm'] - 2.0
            ).clip(lower=36.0)
        self.couverture_2 = peut_reduire_puissance.mean()

        # STRATÉGIE 3 — Free cooling
        peut_free_cooling = (
            (self._get_col(df_out, 'temperature_ambiante', 30)  < 22.0) &
            (self._get_col(df_out, 'vitesse_vent_ms', 0)         > 3.0)  &
            (self._get_col(df_out, 'humidite_relative_pct', 50) < 75.0) &
            (self._get_col(df_out, 'efficacite_free_cooling', 0) > 0.50)
        )
        masque_fc_seul = peut_free_cooling & (df_out['action_proposee'] == 'aucune_action')
        df_out.loc[masque_fc_seul, 'action_proposee'] = 'free_cooling'
        df_out.loc[peut_free_cooling, 'economie_estimee_kwh'] = (
            df_out.loc[peut_free_cooling, 'economie_estimee_kwh'] +
            df_out.loc[peut_free_cooling, 'consommation_kwh'] * 0.15
        )
        self.couverture_3 = peut_free_cooling.mean()

        # STRATÉGIE 4 — Mode éco calendaire
        peut_eco_calendaire = (
            (qos_ok == 1) &
            (
                (self._get_col(df_out, 'est_weekend', 0) == 1) |
                (self._get_col(df_out, 'est_ferie', 0)   == 1)
            ) &
            (df_out['heure'].between(2, 6)) &
            (self._get_col(df_out, 'score_qos', 1)  > 0.80) &
            (self._get_col(df_out, 'taux_charge_data', 0) < 0.13) &
            (df_out['action_proposee'] == 'aucune_action')
        )
        df_out.loc[peut_eco_calendaire, 'action_proposee']      = 'eco_calendaire'
        df_out.loc[peut_eco_calendaire, 'economie_estimee_kwh'] = (
            df_out.loc[peut_eco_calendaire, 'consommation_kwh'] * 0.12
        )
        self.couverture_4 = peut_eco_calendaire.mean()

        # STRATÉGIE 5 — Alerte saturation 2G
        alerte_2g = (
            (df_out['technologie'] == '2G') &
            (self._get_col(df_out, 'taux_charge_voix', 0) > 0.80) &
            (self._get_col(df_out, 'score_qos', 1)         < self.qos_seuil)
        )
        df_out.loc[alerte_2g, 'action_proposee']            = 'alerte_saturation_voix'
        df_out.loc[alerte_2g, 'economie_estimee_kwh']       = 0.0
        df_out.loc[alerte_2g, 'optimisation_qos_autorisee'] = 0
        df_out.loc[alerte_2g, 'priorite_intervention']      = 'Haute'
        self.couverture_5 = alerte_2g.mean()
        return df_out

    def calculer_economies(self, df_in: pd.DataFrame) -> dict:
        baseline = df_in['consommation_kwh'].sum()
        actions  = ['sleep_mode_secteur', 'reduction_puissance', 'free_cooling', 'eco_calendaire']
        noms     = {
            'sleep_mode_secteur' : '1 — Sleep mode secteur',
            'reduction_puissance': '2 — Réduction puissance',
            'free_cooling'       : '3 — Free cooling',
            'eco_calendaire'     : '4 — Mode éco calendaire',
        }
        eco = {}; total_kwh = 0.0
        for action in actions:
            mask  = df_in['action_proposee'] == action
            e_kwh = df_in.loc[mask, 'economie_estimee_kwh'].sum()
            e_pct = e_kwh / baseline * 100 if baseline > 0 else 0
            total_kwh += e_kwh
            eco[noms[action]] = {
                'economie_kwh'  : round(e_kwh, 2),
                'economie_pct'  : round(e_pct, 2),
                'co2_evite_t'   : round(e_kwh * FACTEUR_CO2_TN / 1000, 2),
                'economie_dt'   : round(e_kwh * PRIX_KWH_TN, 2),
                'n_eligible'    : int(mask.sum()),
                'couverture_pct': round(mask.mean() * 100, 1),
                'objectif_10pct': e_pct >= 10,
            }
        total_pct = total_kwh / baseline * 100 if baseline > 0 else 0
        eco['5 — Combinée (1+2+3+4)'] = {
            'economie_kwh'  : round(total_kwh, 2),
            'economie_pct'  : round(total_pct, 2),
            'co2_evite_t'   : round(total_kwh * FACTEUR_CO2_TN / 1000, 2),
            'economie_dt'   : round(total_kwh * PRIX_KWH_TN, 2),
            'objectif_10pct': total_pct >= 10,
        }
        n_alerte = (df_in['action_proposee'] == 'alerte_saturation_voix').sum()
        eco['5 — Alerte saturation 2G (surveillance)'] = {
            'n_observations': int(n_alerte),
            'couverture_pct': round(n_alerte / len(df_in) * 100, 1),
            'economie_kwh'  : 0,
            'action'        : 'migration_3G_4G_ou_extension_capacite',
        }
        eco['baseline_kwh']   = round(baseline, 2)
        eco['baseline_co2_t'] = round(baseline * FACTEUR_CO2_TN / 1000, 2)
        eco['baseline_dt']    = round(baseline * PRIX_KWH_TN, 2)
        return eco


# Validation des constantes physiques (placée ICI, après la classe) 
_eco  = StrategieOptimisation.ECO_SLEEP_PAR_TECHNO
_ord  = StrategieOptimisation._ORDRE_TECHNO

assert all(_eco[t] > 0 for t in _ord), \
    "ECO_SLEEP_PAR_TECHNO : toutes les valeurs doivent être > 0"

assert all(_eco[_ord[i]] <= _eco[_ord[i + 1]] for i in range(len(_ord) - 1)), \
    "ECO_SLEEP_PAR_TECHNO : doit être croissant de 2G à 4G+"

print("Validation ECO_SLEEP_PAR_TECHNO : OK")


strat = StrategieOptimisation(qos_seuil=qos_seuil)
print(f'StrategieOptimisation instanciée avec qos_seuil={strat.qos_seuil}')
print('Application des 5 stratégies d\'optimisation (dataset 2024)...')
df_2024   = strat.appliquer_strategies(df_2024)
economies = strat.calculer_economies(df_2024)

print()
print('  RÉSULTATS DES STRATÉGIES D\'OPTIMISATION (2024)')
print(f"  Baseline : {economies['baseline_kwh']:>12,.0f} kWh")
print(f"  CO2      : {economies['baseline_co2_t']:>12,.1f} tonnes/an")
print(f"  Coût     : {economies['baseline_dt']:>12,.0f} DT/an")
print()
print(f"  {'Stratégie':<35} {'Économie kWh':>14} {'%':>7} {'Éligible':>10} {'Objectif 10%':>14}")
print('  ' + '-'*85)
for nom, r in economies.items():
    if not isinstance(r, dict) or 'economie_pct' not in r: continue
    ok  = '[OK]    ' if r.get('objectif_10pct') else '         '
    n_e = r.get('n_eligible', r.get('n_observations', '-'))
    print(f"  {ok} {nom:<33} {r['economie_kwh']:>12,.0f} {r['economie_pct']:>7.1f}% {str(n_e):>10}")
print()
print('Distribution des actions proposées :')
print(df_2024['action_proposee'].value_counts().to_string())
```

## Phase RL — Apprentissage par Renforcement (7 agents)

**Objectif** : apprendre une politique de décision optimale sans encoder les seuils à la main.
7 algorithmes tabulaires sont comparés sur le même environnement BTS, avec les mêmes hyperparamètres.

| Composant RL | Définition dans ce projet |
|---|---|
| **Environnement** | `BTSEnvironnement` — historique `df_2024` comme simulateur |
| **État** | (bin_qos, bin_charge_data, bin_charge_voix, bin_heure, qos_ok) → 288 états théoriques |
| **Actions** | 5 : aucune, sleep, réduction puissance, free cooling, éco calendaire |
| **Récompense** | +économie_kWh si valide ; −30% conso si violation QoS ; **0 si aucune_action** |


```python
# CELLULE 3b — Environnement de simulation BTS (MDP temporel)
 
class BTSEnvironnement:
    ACTIONS = {
        0: 'aucune_action',
        1: 'sleep_mode_secteur',
        2: 'reduction_puissance',
        3: 'free_cooling',
        4: 'eco_calendaire',
    }
    ECO_FRACTION = {0: 0.00, 1: 0.20, 2: 0.08, 3: 0.15, 4: 0.12}
    QOS_MIN      = {0: None, 1: 0.80, 2: 0.75, 3: None, 4: 0.80}
    PENALITE     = 0.30
    LAMBDA_QOS   = 5.0
 
    def __init__(self, df_data: pd.DataFrame, seed: int = 42):
        self.df        = df_data.reset_index(drop=True)
        self.n_actions = len(self.ACTIONS)
        self._rng      = random.Random(seed)
        self._pos      = 0
        self._indices  = []
        self._qos_carry = 1.0
 
        counts = self.df['station_id'].value_counts()
        self._stations_valides = counts[counts >= 1].index.tolist()
        if not self._stations_valides:
            raise ValueError("Aucune station valide dans df_data.")
 
    def _bin_qos(self, v):
        if v < 0.40: return 0
        if v < 0.60: return 1
        if v < 0.80: return 2
        return 3
 
    def _bin_charge(self, v):
        if v < 0.30: return 0
        if v < 0.60: return 1
        return 2
 
    def _bin_heure(self, h):
        if h <= 5:  return 0
        if h <= 11: return 1
        if h <= 17: return 2
        return 3
 
    def _get_state(self, row, qos_override=None) -> tuple:
        qos = qos_override if qos_override is not None else float(row.get('score_qos', 1.0))
        return (
            self._bin_qos(qos),
            self._bin_charge(float(row.get('taux_charge_data', 0.0))),
            self._bin_charge(float(row.get('taux_charge_voix', 0.0))),
            self._bin_heure(int(row.get('heure', 12))),
            int(row.get('optimisation_qos_autorisee', 1)),
        )
 
    def reset(self):
        """Épisode = 24h consécutives sur une station valide aléatoire."""
        station = self._rng.choice(self._stations_valides)
        idx_station = self.df[self.df['station_id'] == station].index.tolist()
 
        max_start = max(0, len(idx_station) - 24)
        start = self._rng.randint(0, max_start) if max_start > 0 else 0
        self._indices = idx_station[start:start + 24]
        self._pos = 0
        self._qos_carry = float(self.df.loc[self._indices[0], 'score_qos'])
        row = self.df.loc[self._indices[0]]
        return self._get_state(row, qos_override=self._qos_carry), row
 

    def step(self, action: int):
        row       = self.df.loc[self._indices[self._pos]]
        conso     = float(row.get('consommation_kwh', 5.0))
        qos_ok    = int(row.get('optimisation_qos_autorisee', 1))
        score_qos = self._qos_carry
    
        qos_min = self.QOS_MIN[action]
    
        if action == 0:
            # Valoriser le maintien d'une bonne QoS sans action (état stable)
            qos_bonus = max(0.0, score_qos - 0.60) * 0.5  # bonus de stabilité QoS
            reward    = qos_bonus  # ≥ 0, encourage l'inaction dans les bons états
            qos_next  = score_qos  # pas de dégradation
    
        elif qos_min is not None and (score_qos < qos_min or qos_ok == 0):
            ecart_qos = max(0.0, qos_min - score_qos)
            reward    = -conso * self.PENALITE - self.LAMBDA_QOS * ecart_qos
            qos_next  = max(0.0, score_qos - QOS_DELTA_ANOMALIE)
    
        else:
            eco_kwh  = conso * self.ECO_FRACTION[action]
            reward   = eco_kwh
            qos_next = min(1.0, score_qos + QOS_DELTA_NORMAL)
    
        self._qos_carry = qos_next
        self._pos += 1
        done = self._pos >= len(self._indices)
    
        if done:
            next_state, next_row = None, None
        else:
            next_row   = self.df.loc[self._indices[self._pos]]
            next_state = self._get_state(next_row, qos_override=self._qos_carry)
    
        return next_state, reward, done, {
            'action_nom'  : self.ACTIONS[action],
            'score_qos'   : score_qos,
            'qos_next'    : qos_next,
            'economie_kwh': conso * self.ECO_FRACTION[action] if reward > 0 and action != 0 else 0.0,
            'violation'   : 1 if reward < 0 else 0,
        }
 
 
env = BTSEnvironnement(df_2024, seed=42)
print('Environnement RL MDP initialisé')
print(f'  Stations valides  : {len(env._stations_valides)} / {df_2024["station_id"].nunique()}')
print(f'  Épisodes          : 24h consécutives par station (MDP temporel)')
print(f'  Espace états      : 4×3×3×4×2 = {4*3*3*4*2} états')
print(f'  Actions           : {list(env.ACTIONS.values())}')
print(f'  Dynamique QoS     : carry entre steps (dégradation/récupération)')
print(f'  Récompense        : eco_kwh - λ_qos × écart_QoS  (λ={env.LAMBDA_QOS})')
```

```python
# CELLULE 3b2 — 7 Agents RL tabulaires

def get_valid_actions(sq, qok, qos_min_list, n_actions):
    va = [a for a in range(n_actions)
          if qos_min_list[a] is None or (int(qok) == 1 and float(sq) >= qos_min_list[a])]
    return va if va else [0]


class BaseAgent:
    NAME = "Base"
    REF  = ""

    def __init__(self, n_actions, alpha=0.15, gamma=0.95,
                 epsilon=1.0, epsilon_min=0.05, n_episodes=100, seed=42, **kwargs):
        self.n_actions     = n_actions
        self.alpha         = alpha
        self.gamma         = gamma
        self.epsilon       = epsilon
        self.epsilon_min   = epsilon_min
        self.epsilon_decay = epsilon_min ** (1.0 / max(n_episodes, 1))
        self.Q    = defaultdict(lambda: np.zeros(n_actions))
        self._rng = np.random.RandomState(seed)
        self.hist_rewards    = []
        self.hist_epsilon    = []
        self.hist_eco_moy    = []
        self.hist_violations = []
        self.hist_q_max      = []

    def choose_action(self, state, valid_actions):
        if self._rng.rand() < self.epsilon:
            return valid_actions[self._rng.randint(len(valid_actions))]
        q = self.Q[state]
        return valid_actions[int(np.argmax(q[valid_actions]))]

    def greedy_action(self, state, valid_actions):
        q = self.Q[state]
        return valid_actions[int(np.argmax(q[valid_actions]))]

    def update(self, s, a, r, s_next, va_next, done, **kwargs):
        raise NotImplementedError

    def decay_epsilon(self):
        self.epsilon = max(self.epsilon_min, self.epsilon * self.epsilon_decay)

    def get_q_max(self):
        vals = list(self.Q.values())
        return float(max(np.max(v) for v in vals)) if vals else 0.0


class AgentQLearning(BaseAgent):
    NAME = "Q-Learning"
    REF  = "Watkins & Dayan (1992)"

    def update(self, s, a, r, s_next, va_next, done, **kwargs):
        q_target = r if done else r + self.gamma * np.max(self.Q[s_next][va_next])
        self.Q[s][a] += self.alpha * (q_target - self.Q[s][a])


class AgentSARSA(BaseAgent):
    NAME = "SARSA"
    REF  = "Rummery & Niranjan (1994)"

    def update(self, s, a, r, s_next, va_next, done, **kwargs):
        a_next   = kwargs.get('a_next', None)
        q_target = r if (done or a_next is None) else r + self.gamma * self.Q[s_next][a_next]
        self.Q[s][a] += self.alpha * (q_target - self.Q[s][a])


class AgentDoubleQLearning(BaseAgent):
    NAME = "Double Q-Learning"
    REF  = "van Hasselt (2010)"

    def __init__(self, n_actions, alpha=0.15, gamma=0.95,
                 epsilon=1.0, epsilon_min=0.05, n_episodes=100, seed=42, **kwargs):
        # super().__init__ appelle self.Q = defaultdict(...)
        # Le setter doit être silencieux à ce stade (Q1 n'existe pas encore)
        super().__init__(n_actions, alpha=alpha, gamma=gamma,
                         epsilon=epsilon, epsilon_min=epsilon_min,
                         n_episodes=n_episodes, seed=seed)
        # Q1 et Q2 initialisés APRÈS super() → le setter devient bloquant dès ici
        self.Q1 = defaultdict(lambda: np.zeros(n_actions))
        self.Q2 = defaultdict(lambda: np.zeros(n_actions))

    @property
    def Q(self):
        """Retourne la moyenne (Q1+Q2)/2 sous forme de dict."""
        all_s = set(self.Q1.keys()) | set(self.Q2.keys())
        return {s: (self.Q1[s] + self.Q2[s]) / 2 for s in all_s}

    @Q.setter
    def Q(self, value):
        # Logique du setter intelligent 
        # Pendant super().__init__(), Q1 n'existe pas encore → on laisse passer
        # silencieusement (c'est la phase d'initialisation de BaseAgent).
        # Une fois Q1 initialisé, toute écriture externe sur .Q est bloquée.
        if not hasattr(self, 'Q1'):
            return   # init phase : ignore silencieusement l'assignation de BaseAgent
        raise AttributeError(
            "AgentDoubleQLearning : Q est une propriété calculée (Q1+Q2)/2. "
            "Écrivez directement dans self.Q1 ou self.Q2."
        )

    def choose_action(self, state, valid_actions):
        if self._rng.rand() < self.epsilon:
            return valid_actions[self._rng.randint(len(valid_actions))]
        q = self.Q1[state] + self.Q2[state]
        return valid_actions[int(np.argmax(q[valid_actions]))]

    def greedy_action(self, state, valid_actions):
        q = self.Q1[state] + self.Q2[state]
        return valid_actions[int(np.argmax(q[valid_actions]))]

    def update(self, s, a, r, s_next, va_next, done, **kwargs):
        if self._rng.rand() < 0.5:
            a_star   = va_next[int(np.argmax(self.Q1[s_next][va_next]))]
            q_target = r if done else r + self.gamma * self.Q2[s_next][a_star]
            self.Q1[s][a] += self.alpha * (q_target - self.Q1[s][a])
        else:
            a_star   = va_next[int(np.argmax(self.Q2[s_next][va_next]))]
            q_target = r if done else r + self.gamma * self.Q1[s_next][a_star]
            self.Q2[s][a] += self.alpha * (q_target - self.Q2[s][a])

    def get_q_max(self):
        all_s = set(self.Q1.keys()) | set(self.Q2.keys())
        if not all_s: return 0.0
        return float(max(np.max((self.Q1[s] + self.Q2[s]) / 2) for s in all_s))


class AgentExpectedSARSA(BaseAgent):
    NAME = "Expected SARSA"
    REF  = "van Seijen et al. (2009)"

    def update(self, s, a, r, s_next, va_next, done, **kwargs):
        if done:
            q_target = r
        else:
            q_next   = self.Q[s_next]
            a_greedy = va_next[int(np.argmax(q_next[va_next]))]
            expected = ((1.0 - self.epsilon) * q_next[a_greedy] +
                        self.epsilon * float(np.mean(q_next[va_next])))
            q_target = r + self.gamma * expected
        self.Q[s][a] += self.alpha * (q_target - self.Q[s][a])


class AgentQLearningUCB(BaseAgent):
    NAME = "Q-Learning UCB"
    REF  = "Auer et al. (2002)"

    def __init__(self, n_actions, alpha=0.15, gamma=0.95,
                 c=1.5, n_episodes=100, seed=42, **kwargs):
        super().__init__(n_actions, alpha=alpha, gamma=gamma,
                         epsilon=0.0, epsilon_min=0.0,
                         n_episodes=n_episodes, seed=seed)
        self.c = c
        self.N = defaultdict(lambda: np.ones(n_actions))
        self.t = 1

    def choose_action(self, state, valid_actions):
        self.t += 1
        q   = self.Q[state]
        n   = self.N[state]
        ucb = q + self.c * np.sqrt(np.log(self.t) / n)
        return valid_actions[int(np.argmax(ucb[valid_actions]))]

    def greedy_action(self, state, valid_actions):
        q = self.Q[state]
        return valid_actions[int(np.argmax(q[valid_actions]))]

    def update(self, s, a, r, s_next, va_next, done, **kwargs):
        self.N[s][a] += 1
        q_target = r if done else r + self.gamma * np.max(self.Q[s_next][va_next])
        self.Q[s][a] += self.alpha * (q_target - self.Q[s][a])

    def decay_epsilon(self):
        pass   # UCB n'utilise pas epsilon


class AgentSARSALambda(BaseAgent):
    NAME = "SARSA(λ)"
    REF  = "Sutton (1988)"

    def __init__(self, n_actions, alpha=0.15, gamma=0.95, lam=0.7,
                 epsilon=1.0, epsilon_min=0.05, n_episodes=100, seed=42, **kwargs):
        super().__init__(n_actions, alpha=alpha, gamma=gamma,
                         epsilon=epsilon, epsilon_min=epsilon_min,
                         n_episodes=n_episodes, seed=seed)
        self.lam = lam
        self.E   = defaultdict(lambda: np.zeros(n_actions))

    def new_episode(self):
        self.E.clear()

    def update(self, s, a, r, s_next, va_next, done, **kwargs):
        a_next = kwargs.get('a_next', None)
        delta  = (r - self.Q[s][a]) if (done or a_next is None) else (
                  r + self.gamma * self.Q[s_next][a_next] - self.Q[s][a])
        self.E[s][a] += 1.0
        for state in list(self.E.keys()):
            self.Q[state] += self.alpha * delta * self.E[state]
            self.E[state] *= self.gamma * self.lam
        if done:
            self.E.clear()

class AgentMonteCarlo(BaseAgent):
    NAME = "Monte Carlo"
    REF  = "Sutton & Barto (1998)"

    def __init__(self, *args, **kwargs):
        super().__init__(*args, **kwargs)
        self._buffer  = []
        self._returns = defaultdict(list)

    def update(self, s, a, r, s_next, va_next, done, **kwargs):
        self._buffer.append((s, a, r))
        if done:
            self._mc_update()
            # NOTE : decay_epsilon() est géré DANS la boucle d'entraînement
            # via le flag is_mc=False dans train_all_agents, comme tous les autres agents.
            # NE PAS appeler decay_epsilon() ici.

    def _mc_update(self):
        G, visited = 0.0, set()
        for s, a, r in reversed(self._buffer):
            G = r + self.gamma * G
            if (s, a) not in visited:
                visited.add((s, a))
                self._returns[(s, a)].append(G)
                self.Q[s][a] = float(np.mean(self._returns[(s, a)]))
        self._buffer.clear()


ALL_AGENT_CLASSES = [
    AgentQLearning, AgentSARSA, AgentDoubleQLearning,
    AgentExpectedSARSA, AgentQLearningUCB, AgentSARSALambda, AgentMonteCarlo
]
```

```python
# CELLULE 3c — Entraînement des 7 agents RL

# On trie par mois pour garder la chronologie (mois 1-9 → train, 10-12 → test)
df_2024_sorted = df_2024.sort_values(['mois', 'heure']).reset_index(drop=True)
n_total = len(df_2024_sorted)
n_train = int(n_total * 0.75)

df_train_rl = df_2024_sorted.iloc[:n_train].copy().reset_index(drop=True)
df_test_rl  = df_2024_sorted.iloc[n_train:].copy().reset_index(drop=True)

print('  SPLIT TRAIN / TEST (chronologique)')
print(f'  Total   : {n_total:>10,} observations')
print(f'  Train   : {len(df_train_rl):>10,} obs. | {df_train_rl["station_id"].nunique()} stations')
print(f'  Test    : {len(df_test_rl):>10,}  obs. | {df_test_rl["station_id"].nunique()} stations')
if 'mois' in df_train_rl.columns:
    print(f'  Mois train : {df_train_rl["mois"].min()}-{df_train_rl["mois"].max()}')
    print(f'  Mois test  : {df_test_rl["mois"].min()}-{df_test_rl["mois"].max()}')
print()

env_train = BTSEnvironnement(df_train_rl, seed=SEED)
env_test  = BTSEnvironnement(df_test_rl,  seed=SEED)

print(f'env_train : {len(env_train._stations_valides)} stations valides')
print(f'env_test  : {len(env_test._stations_valides)} stations valides')
print()

N_EPISODES   = 1000
ALPHA        = 0.15
GAMMA        = 0.95
EPS_MIN      = 0.05

QOS_MIN_LIST = [env_train.QOS_MIN[a] for a in range(env_train.n_actions)]
ECO_FRAC_ARR = np.array([env_train.ECO_FRACTION[a] for a in range(env_train.n_actions)],
                          dtype=np.float32)


def train_all_agents(env_ref):
    """Entraîne les 7 agents sur env_ref (doit être env_train)."""
    agents_dict = {
        'Q-Learning'        : AgentQLearning(env_ref.n_actions,       alpha=ALPHA, gamma=GAMMA, n_episodes=N_EPISODES, seed=SEED),
        'SARSA'             : AgentSARSA(env_ref.n_actions,           alpha=ALPHA, gamma=GAMMA, n_episodes=N_EPISODES, seed=SEED),
        'Double Q-Learning' : AgentDoubleQLearning(env_ref.n_actions,  alpha=ALPHA, gamma=GAMMA, n_episodes=N_EPISODES, seed=SEED),
        'Expected SARSA'    : AgentExpectedSARSA(env_ref.n_actions,    alpha=ALPHA, gamma=GAMMA, n_episodes=N_EPISODES, seed=SEED),
        'Q-Learning UCB'    : AgentQLearningUCB(env_ref.n_actions,     alpha=ALPHA, gamma=GAMMA, c=1.5, n_episodes=N_EPISODES, seed=SEED),
        'SARSA(λ)'          : AgentSARSALambda(env_ref.n_actions,      alpha=ALPHA, gamma=GAMMA, lam=0.7, n_episodes=N_EPISODES, seed=SEED),
        'Monte Carlo'       : AgentMonteCarlo(env_ref.n_actions,       alpha=ALPHA, gamma=GAMMA, n_episodes=N_EPISODES, seed=SEED),
    }

    for name, agent in agents_dict.items():
        is_sarsa = isinstance(agent, (AgentSARSA, AgentSARSALambda))
        is_mc    = isinstance(agent, AgentMonteCarlo)

        print(f'  [{name:<22}] ', end='', flush=True)
        t0 = time.time()

        for ep in range(1, N_EPISODES + 1):

            state, _ = env_ref.reset()
            indices  = env_ref._indices
            tot_r = tot_eco = tot_viol = 0.0

            if isinstance(agent, AgentSARSALambda):
                agent.new_episode()

            a_current = None
            if is_sarsa:
                va0 = get_valid_actions(
                    env_ref._qos_carry,
                    int(env_ref.df.loc[indices[0], 'optimisation_qos_autorisee']),
                    QOS_MIN_LIST, env_ref.n_actions
                )
                a_current = agent.choose_action(state, va0)

            done = False
            while not done:
                idx = env_ref._indices[env_ref._pos] if env_ref._pos < len(env_ref._indices) else env_ref._indices[-1]
                sq  = env_ref._qos_carry
                qok = int(env_ref.df.loc[idx, 'optimisation_qos_autorisee'])
                va  = get_valid_actions(sq, qok, QOS_MIN_LIST, env_ref.n_actions)
                a   = a_current if is_sarsa else agent.choose_action(state, va)

                next_state, reward, done, info = env_ref.step(a)

                if not done:
                    va_n = get_valid_actions(
                        env_ref._qos_carry,
                        int(env_ref.df.loc[env_ref._indices[env_ref._pos], 'optimisation_qos_autorisee']),
                        QOS_MIN_LIST, env_ref.n_actions
                    )
                    a_next = agent.choose_action(next_state, va_n) if is_sarsa else None
                else:
                    va_n = va; a_next = None; next_state = state

                agent.update(state, a, reward, next_state, va_n, done, a_next=a_next)

                if is_sarsa:
                    a_current = a_next

                state    = next_state if next_state is not None else state
                tot_r    += reward
                tot_eco  += info['economie_kwh']
                tot_viol += info['violation']

            agent.decay_epsilon()
            agent.hist_rewards.append(tot_r)
            agent.hist_epsilon.append(agent.epsilon)
            agent.hist_eco_moy.append(tot_eco / max(len(indices), 1))
            agent.hist_violations.append(int(tot_viol))
            agent.hist_q_max.append(agent.get_q_max())

        elapsed = time.time() - t0
        print(f'OK  ({elapsed:.1f}s) | ε={agent.epsilon:.3f} | '
              f'éco={np.mean(agent.hist_eco_moy[-20:]):.4f} kWh/step | '
              f'Q_max={agent.hist_q_max[-1]:.3f}')

    return agents_dict


print(f'Entraînement de 7 agents — {N_EPISODES} épisodes × 24 steps/épisode')
print(f'Jeu d\'entraînement : {len(df_train_rl):,} obs. ({len(env_train._stations_valides)} stations)')
print(f'Hyperparamètres   : α={ALPHA}, γ={GAMMA}, ε_min={EPS_MIN}')
print(f'Justification     : α=0.15 → convergence stable sur épisodes courts (24 steps)')
print(f'                    γ=0.95 → horizon effectif ~20 steps (1/(1-γ))')
print(f'                    λ=0.70 → SARSA(λ) : compromis éligibilité/stabilité (Sutton 1988)')
print(f'                    c=1.50 → UCB : balance exploration/exploitation (Auer 2002)')
print()

# Entraînement sur env_train uniquement
agents_trained = train_all_agents(env_train)
print('\nTous les agents entraînés avec succès sur le jeu d\'entraînement.')
```

```python
# CELLULE 3d — Évaluation & comparaison des 7 agents vs Règles Expertes

env_eval = BTSEnvironnement(df_test_rl, seed=SEED)


def evaluate_agent_vec(agent, env_ref, df_eval):
    """
    Inférence greedy vectorisée (ε=0) sur df_eval.
    NOTE MÉTHODOLOGIQUE : évaluation statique (sans QoS carry inter-observations).
    Cette approximation est conservatrice — elle sous-estime légèrement les économies
    réelles que produirait un déploiement épisodique complet.
    Pour une évaluation exacte, utiliser evaluate_agent_episodique() ci-dessous.
    """
    eps_save      = agent.epsilon
    agent.epsilon = 0.0

    sq_arr    = df_eval['score_qos'].values.astype(np.float32)
    qok_arr   = df_eval['optimisation_qos_autorisee'].values.astype(np.int8)
    con_arr   = df_eval['consommation_kwh'].values.astype(np.float32)
    bqos_arr  = np.digitize(sq_arr, [0.40, 0.60, 0.80]).astype(np.int8)
    bdata_arr = np.digitize(df_eval['taux_charge_data'].values.astype(np.float32),
                             [0.30, 0.60]).astype(np.int8)
    bvoix_arr = np.digitize(df_eval['taux_charge_voix'].values.astype(np.float32),
                             [0.30, 0.60]).astype(np.int8)
    bheur_arr = np.digitize(df_eval['heure'].values.astype(np.int8),
                             [5, 11, 17]).astype(np.int8)

    n           = len(df_eval)
    actions_out = ['aucune_action'] * n
    eco_out     = np.zeros(n, dtype=np.float32)
    viol_out    = np.zeros(n, dtype=np.int8)

    for i in range(n):
        state = (int(bqos_arr[i]), int(bdata_arr[i]),
                 int(bvoix_arr[i]), int(bheur_arr[i]), int(qok_arr[i]))
        sq  = float(sq_arr[i])
        qok = int(qok_arr[i])
        va  = get_valid_actions(sq, qok, QOS_MIN_LIST, env_ref.n_actions)

        action = agent.greedy_action(state, va)

        qm = env_ref.QOS_MIN[action]
        if action != 0 and qm is not None and (sq < qm or qok == 0):
            viol_out[i] = 1
        else:
            actions_out[i] = env_ref.ACTIONS[action]
            eco_out[i]     = con_arr[i] * env_ref.ECO_FRACTION[action]

    agent.epsilon = eps_save
    return actions_out, eco_out.tolist(), viol_out.tolist()


def evaluate_agent_episodique(agent, env_ref, n_episodes=100):
    """
    Évaluation réaliste avec QoS carry propagé entre steps (vs évaluation statique).
    Reflète fidèlement un déploiement épisodique complet.
    """
    eps_save      = agent.epsilon
    agent.epsilon = 0.0
    total_eco = total_viols = total_steps = 0
    is_sarsa = isinstance(agent, (AgentSARSA, AgentSARSALambda))

    for _ in range(n_episodes):
        state, _ = env_ref.reset()
        done      = False

        if is_sarsa:
            idx0 = env_ref._indices[0]
            va0  = get_valid_actions(env_ref._qos_carry,
                                     int(env_ref.df.loc[idx0, 'optimisation_qos_autorisee']),
                                     QOS_MIN_LIST, env_ref.n_actions)
            a_cur = agent.greedy_action(state, va0)
        else:
            a_cur = None

        while not done:
            idx = (env_ref._indices[env_ref._pos]
                   if env_ref._pos < len(env_ref._indices)
                   else env_ref._indices[-1])
            sq  = env_ref._qos_carry
            qok = int(env_ref.df.loc[idx, 'optimisation_qos_autorisee'])
            va  = get_valid_actions(sq, qok, QOS_MIN_LIST, env_ref.n_actions)

            action = a_cur if is_sarsa else agent.greedy_action(state, va)

            state, reward, done, info = env_ref.step(action)
            total_eco   += info['economie_kwh']
            total_viols += info['violation']
            total_steps += 1

            if is_sarsa and not done:
                idx_n = env_ref._indices[env_ref._pos]
                va_n  = get_valid_actions(env_ref._qos_carry,
                                          int(env_ref.df.loc[idx_n, 'optimisation_qos_autorisee']),
                                          QOS_MIN_LIST, env_ref.n_actions)
                a_cur = agent.greedy_action(state, va_n)

    agent.epsilon = eps_save
    eco_moy_step  = total_eco / max(total_steps, 1)
    return eco_moy_step, total_viols, total_steps


# Justification empirique de alpha 
print('Sensibilité à alpha — Q-Learning, 100 épisodes, env_train :')
print(f'  {"alpha":>6}  {"éco%":>8}')
print('  ' + '─'*20)
_env_tmp = BTSEnvironnement(df_train_rl, seed=SEED)
for _alpha_test in [0.05, 0.10, 0.15, 0.20, 0.30]:
    _ag_tmp = AgentQLearning(_env_tmp.n_actions, alpha=_alpha_test,
                              gamma=0.95, n_episodes=100, seed=SEED)
    for _ in range(100):
        _s, _ = _env_tmp.reset()
        _done = False
        while not _done:
            _idx = (_env_tmp._indices[_env_tmp._pos]
                    if _env_tmp._pos < len(_env_tmp._indices)
                    else _env_tmp._indices[-1])
            _va  = get_valid_actions(_env_tmp._qos_carry,
                                     int(_env_tmp.df.loc[_idx, 'optimisation_qos_autorisee']),
                                     QOS_MIN_LIST, _env_tmp.n_actions)
            _a   = _ag_tmp.choose_action(_s, _va)
            _sn, _r, _done, _ = _env_tmp.step(_a)
            _ag_tmp.update(_s, _a, _r, _sn if _sn is not None else _s, _va, _done)
            _s   = _sn if _sn is not None else _s
        _ag_tmp.decay_epsilon()
    _env_eval_tmp = BTSEnvironnement(df_test_rl, seed=SEED)
    _, _ecos_tmp, _ = evaluate_agent_vec(_ag_tmp, _env_eval_tmp, df_test_rl)
    _eco_tmp = sum(_ecos_tmp) / df_test_rl['consommation_kwh'].sum() * 100
    marker = ' ← choix retenu' if _alpha_test == 0.15 else ''
    print(f'  {_alpha_test:>6.2f}  {_eco_tmp:>7.2f}%{marker}')
del _env_tmp, _env_eval_tmp, _ag_tmp, _ecos_tmp
print()


baseline_test       = df_test_rl['consommation_kwh'].sum()
eco_regles_test_kwh = df_test_rl['economie_estimee_kwh'].sum()
eco_regles_test_pct = eco_regles_test_kwh / baseline_test * 100

print('Inférence greedy sur df_test_rl (données NON VUES pendant l\'entraînement)...\n')

rl_results = {}
for name, agent in agents_trained.items():
    t0 = time.time()

    acts, ecos, viols = evaluate_agent_vec(agent, env_eval, df_test_rl)

    eco_kwh = sum(ecos)
    eco_pct = eco_kwh / baseline_test * 100
    n_viols = sum(viols)

    rl_results[name] = {
        'agent'    : agent,
        'eco_kwh'  : eco_kwh,
        'eco_pct'  : eco_pct,
        'co2_t'    : eco_kwh * FACTEUR_CO2_TN / 1000,
        'eco_dt'   : eco_kwh * PRIX_KWH_TN,
        'n_viols'  : n_viols,
        'pct_viols': n_viols / len(df_test_rl) * 100,
        'actions'  : acts,
        'ecos'     : ecos,
        'viols'    : viols,
    }

    col = f'eco_rl_{name.lower().replace(" ","_").replace("(","").replace(")","").replace("λ","lam")}_kwh'
    df_test_rl[col] = ecos
    df_test_rl[f'action_rl_{name.split()[0].lower()}'] = acts
    print(f'  {name:<22}: éco={eco_pct:5.2f}% | viols={n_viols:6,} ({n_viols/len(df_test_rl)*100:.2f}%) | {time.time()-t0:.1f}s')

SEUIL_ECO_MIN_PCT = 2.0

agents_eligibles = [n for n in rl_results if rl_results[n]['eco_pct'] >= SEUIL_ECO_MIN_PCT]
if not agents_eligibles:
    print(f'\nATTENTION : aucun agent n\'atteint le seuil de {SEUIL_ECO_MIN_PCT}% — seuil abaissé à 0%')
    agents_eligibles = list(rl_results.keys())

best_name = min(
    agents_eligibles,
    key=lambda n: (rl_results[n]['n_viols'], -rl_results[n]['eco_kwh'])
)
top_eco_name = max(rl_results, key=lambda n: rl_results[n]['eco_kwh'])
best_agent   = rl_results[best_name]['agent']

eco_rl_kwh   = rl_results[best_name]['eco_kwh']
eco_rl_pct   = rl_results[best_name]['eco_pct']
n_violations = rl_results[best_name]['n_viols']

# Cellule 4 utilise eco_regles_pct mais seul eco_regles_test_pct était défini
eco_regles_pct = eco_regles_test_pct

# Propager le meilleur agent sur df_test_rl
df_test_rl['action_rl']       = rl_results[best_name]['actions']
df_test_rl['economie_rl_kwh'] = rl_results[best_name]['ecos']
df_test_rl['violation_rl']    = rl_results[best_name]['viols']

df_2024['action_rl']       = 'aucune_action'
df_2024['economie_rl_kwh'] = 0.0
df_2024['violation_rl']    = 0

# Reconstruction des index originaux dans le même ordre que df_2024_sorted
original_idx_sorted = df_2024.sort_values(['mois', 'heure']).index   # même tri que Cellule 3c
idx_test_original   = original_idx_sorted[n_train:]                  # portion test

assert len(idx_test_original) == len(rl_results[best_name]['actions']), (
    f"Longueur incohérente : idx_test_original={len(idx_test_original)} "
    f"vs actions={len(rl_results[best_name]['actions'])}. "
    f"Vérifier que n_train ({n_train}) est cohérent avec df_2024_sorted."
)

actions_series = pd.Series(rl_results[best_name]['actions'], index=idx_test_original)
ecos_series    = pd.Series(rl_results[best_name]['ecos'],    index=idx_test_original)
viols_series   = pd.Series(rl_results[best_name]['viols'],   index=idx_test_original)

df_2024.loc[idx_test_original, 'action_rl']       = actions_series
df_2024.loc[idx_test_original, 'economie_rl_kwh'] = ecos_series
df_2024.loc[idx_test_original, 'violation_rl']    = viols_series

print(f'\nPropagation meilleur agent ({best_name}) → df_2024 :')
print(f'  Observations mises à jour : {len(idx_test_original):,}')
print(f'  action_rl ≠ aucune_action : {(df_2024["action_rl"] != "aucune_action").sum():,}')
print(f'  economie_rl_kwh > 0       : {(df_2024["economie_rl_kwh"] > 0).sum():,}')

print('\nPropagation colonnes eco_rl par agent → df_2024 :')
for name, r in rl_results.items():
    col = f'eco_rl_{name.lower().replace(" ","_").replace("(","").replace(")","").replace("λ","lam")}_kwh'
    df_2024[col] = 0.0
    ecos_agent = pd.Series(r['ecos'], index=idx_test_original)
    df_2024.loc[idx_test_original, col] = ecos_agent
    print(f'  {col:<55} sum={df_2024[col].sum():,.1f} kWh')

print()
print('   COMPARAISON : RÈGLES EXPERTES vs 7 AGENTS RL  [sur jeu de TEST]')
print(f'  Baseline test consommation 2024 : {baseline_test:>14,.1f} kWh')
print()
print(f'  {"Agent":<24} {"Éco kWh":>12} {"Éco %":>7} {"CO2 évité(t)":>13} {"Éco DT":>10} {"Violations":>11} {"Viol%":>7}')
print('  ' + '─'*90)
print(f'  {"[Règles Expertes]":<24} {eco_regles_test_kwh:>12,.0f} {eco_regles_test_pct:>7.2f}% '
      f'{eco_regles_test_kwh*FACTEUR_CO2_TN/1000:>12.1f}t {eco_regles_test_kwh*PRIX_KWH_TN:>10,.0f} '
      f'{"N/A":>11} {"N/A":>7}')
for name, r in rl_results.items():
    star = ' ★' if name == best_name else '   '
    print(f'  {name+star:<24} {r["eco_kwh"]:>12,.0f} {r["eco_pct"]:>7.2f}% '
          f'{r["co2_t"]:>12.1f}t {r["eco_dt"]:>10,.0f} '
          f'{r["n_viols"]:>11,} {r["pct_viols"]:>7.2f}%')
print()
print(f'  ★ Meilleur agent (sécurité QoS + économies, seuil ≥{SEUIL_ECO_MIN_PCT}%) : {best_name}')
print(f'  ▲ Agent le plus économe                                           : {top_eco_name}')

# Évaluation épisodique du meilleur agent
print('\nÉvaluation épisodique du meilleur agent (QoS carry propagé)...')
eco_ep_step, viols_ep, steps_ep = evaluate_agent_episodique(best_agent, env_eval, n_episodes=200)
eco_ep_kwh = eco_ep_step * steps_ep
eco_ep_pct = eco_ep_kwh / baseline_test * 100
ecart_pct  = eco_ep_pct - eco_rl_pct

print(f'  [{best_name}] éval. statique  : {eco_rl_pct:.2f}%')
print(f'  [{best_name}] éval. épisodique: {eco_ep_pct:.2f}%')
print(f'  Écart (sous-estimation statique) : {ecart_pct:+.2f}%')
print(f'  Violations épisodiques           : {viols_ep:,} / {steps_ep:,} steps ({viols_ep/max(steps_ep,1)*100:.2f}%)')
print()
print('NOTE BASELINE :')
print('  La baseline utilisée = consommation mesurée 2024 SANS optimisation active.')
print('  Elle ne tient pas compte des pratiques manuelles déjà en place.')
print('  Les gains réels en production peuvent différer selon le niveau d\'optimisation existant.')


# Stabilité multi-seeds
print('\nStabilité des économies RL (5 runs, seeds 0-4, évaluation sur df_test_rl) :')
print(f'  {"Agent":<24} {"Éco% moy":>9} {"± std":>7} {"Min":>7} {"Max":>7}')
print('  ' + '─'*58)

for name, AgClass in zip(
    ['Q-Learning', 'SARSA', 'Double Q-Learning', 'Expected SARSA',
     'Q-Learning UCB', 'SARSA(λ)', 'Monte Carlo'],
    ALL_AGENT_CLASSES
):
    runs_eco = []
    for seed in range(5):
        kwargs = dict(n_actions=env_train.n_actions, alpha=ALPHA, gamma=GAMMA,
                      n_episodes=200, seed=seed)
        if name == 'Q-Learning UCB': kwargs['c']   = 1.5
        if name == 'SARSA(λ)':       kwargs['lam'] = 0.7
        ag_tmp = AgClass(**kwargs)

        is_sarsa_tmp = isinstance(ag_tmp, (AgentSARSA, AgentSARSALambda))
        is_mc_tmp    = isinstance(ag_tmp, AgentMonteCarlo)

        for _ in range(50):
            s, _ = env_train.reset()
            done  = False

            if isinstance(ag_tmp, AgentSARSALambda):
                ag_tmp.new_episode()

            a_cur = None
            if is_sarsa_tmp:
                va0 = get_valid_actions(
                    env_train._qos_carry,
                    int(env_train.df.loc[env_train._indices[0], 'optimisation_qos_autorisee']),
                    QOS_MIN_LIST, env_train.n_actions
                )
                a_cur = ag_tmp.choose_action(s, va0)

            while not done:
                idx = (env_train._indices[env_train._pos]
                       if env_train._pos < len(env_train._indices)
                       else env_train._indices[-1])
                va  = get_valid_actions(
                    env_train._qos_carry,
                    int(env_train.df.loc[idx, 'optimisation_qos_autorisee']),
                    QOS_MIN_LIST, env_train.n_actions
                )
                a = a_cur if is_sarsa_tmp else ag_tmp.choose_action(s, va)
                s_next, r, done, _ = env_train.step(a)

                if not done:
                    va_n = get_valid_actions(
                        env_train._qos_carry,
                        int(env_train.df.loc[env_train._indices[env_train._pos],
                                             'optimisation_qos_autorisee']),
                        QOS_MIN_LIST, env_train.n_actions
                    )
                    a_next = ag_tmp.choose_action(s_next, va_n) if is_sarsa_tmp else None
                else:
                    va_n = va; a_next = None; s_next = s

                ag_tmp.update(s, a, r, s_next, va_n, done, a_next=a_next)

                if is_sarsa_tmp:
                    a_cur = a_next
                s = s_next if s_next is not None else s

            if not is_mc_tmp:
                ag_tmp.decay_epsilon()

        _, ecos_r, _ = evaluate_agent_vec(ag_tmp, env_eval, df_test_rl)
        runs_eco.append(sum(ecos_r) / baseline_test * 100)

    m, s_d = np.mean(runs_eco), np.std(runs_eco)
    print(f'  {name:<24} {m:>8.2f}% {s_d:>6.2f}% {min(runs_eco):>6.2f}% {max(runs_eco):>6.2f}%')
```

## Analyse & interprétation des résultats RL

### Comparaison des algorithmes

Les résultats obtenus sur le jeu de **test indépendant** (données non vues pendant l'entraînement)
permettent les observations suivantes :

- **Q-Learning et Expected SARSA** convergent généralement plus vite grâce à leur mise à jour
  *off-policy* qui exploite le maximum de la Q-table, avantageuse sur des épisodes courts (24 steps).

- **SARSA(λ)** bénéficie de la propagation par traces d'éligibilité (λ=0.7), ce qui accélère
  l'attribution de crédit sur plusieurs steps consécutifs — utile pour les stratégies nocturnes
  (sleep mode, éco calendaire) qui s'étalent sur plusieurs heures.

- **Monte Carlo** converge plus lentement car le retour est calculé en fin d'épisode uniquement,
  sans mise à jour intermédiaire. Il devient compétitif après ~500 épisodes.

- **Q-Learning UCB** (c=1.5) explore de manière plus systématique que ε-greedy pur, ce qui
  réduit la variance inter-seeds au prix d'une convergence initiale plus lente.

### Taux de violation QoS

Tous les agents présentent un taux de violation faible grâce au **garde-fou QoS** intégré dans
`get_valid_actions()`. Les violations résiduelles correspondent à des états où le score QoS
était supérieur au seuil requis au moment de la décision, mais a chuté entre la décision et
l'application de l'action (dynamique temporelle non capturable en évaluation statique).

### Comparaison Règles expertes vs RL

Les règles expertes définissent des conditions précises et calibrées (seuils horaires, météo).
Les agents RL apprennent une politique plus souple, capable de s'adapter aux stations atypiques
non couvertes par les règles. La différence de performance reflète le compromis entre
**explicabilité** (règles) et **généralisation** (RL).

### Note sur la méthodologie d'évaluation

L'évaluation vectorisée (`evaluate_agent_vec`) traite chaque observation indépendamment, sans
propager le QoS carry entre steps. Cette approximation conservatrice est cohérente avec un
déploiement réel où chaque décision est prise sur la mesure courante. Elle sous-estime
légèrement les économies réelles d'un déploiement épisodique complet.

```python
# Cellule 5 — Pipeline d'inférence temps réel (avec meilleur agent RL intégré)

class PipelineInferenceTempsReel:
    """
    Pipeline complet d'inférence pour une mesure temps réel (station × heure).
    Intègre le meilleur agent RL en parallèle du moteur de règles.
    """

    def __init__(self, modele_lgbm, modele_anom, encodeurs, quantiles,
                 features, moteur_decision, agent_rl=None, env_rl=None):
        self.modele_lgbm   = modele_lgbm
        self.modele_anom   = modele_anom
        self.encodeurs     = encodeurs
        self.quantiles     = quantiles
        self.features      = features
        self.moteur        = moteur_decision
        self.scaler        = modele_anom['scaler_std']
        self.iso_f         = modele_anom['iso_f']
        self.anom_features = modele_anom['features']
        self.agent_rl      = agent_rl
        self.env_rl        = env_rl

    def _encoder(self, row: dict) -> dict:
        try:
            row['technologie_enc'] = int(self.encodeurs['technologie'].transform([row.get('technologie', '4G')])[0])
            row['type_zone_enc']   = int(self.encodeurs['type_zone'].transform([row.get('type_zone', 'Urbain')])[0])
        except Exception:
            row['technologie_enc'] = 0; row['type_zone_enc'] = 0
        return row

    def _enrichir(self, row: dict) -> dict:
        h = row.get('heure', 12); j = row.get('jour_semaine', 1); m = row.get('mois', 6)
        row['heure_sin'] = np.sin(2*np.pi*h/24); row['heure_cos'] = np.cos(2*np.pi*h/24)
        row['jour_sin']  = np.sin(2*np.pi*j/7);  row['jour_cos']  = np.cos(2*np.pi*j/7)
        row['mois_sin']  = np.sin(2*np.pi*m/12); row['mois_cos']  = np.cos(2*np.pi*m/12)
        row['charge_thermique']     = max(0, row.get('temperature_ambiante', 25) - 28)
        row['charge_par_secteur']   = row.get('charge_cpu_pct', 50) / max(row.get('nb_secteurs_actifs', 1), 1)
        row['efficacite_spectrale'] = (row.get('trafic_data_mbps', 1)+1) / (row.get('puissance_emission_dbm', 43)-30+1)
        row['intensite_trafic']     = row.get('nb_utilisateurs_actifs', 50) / (row.get('trafic_data_mbps', 1)+0.1)
        if 'score_qos' not in row:
            tv = row.get('taux_charge_voix', row.get('trafic_voix_erlang', 0) / 80.0)
            td = row.get('taux_charge_data', row.get('trafic_data_mbps', 0) / 100.0)
            row['score_qos'] = max(0.0, 0.50*(1-td) + 0.30*(1-tv) + 0.20*(1-row.get('charge_cpu_pct',50)/100))
        if 'optimisation_qos_autorisee' not in row:
            tv = row.get('taux_charge_voix', 0); td = row.get('taux_charge_data', 0)
            row['optimisation_qos_autorisee'] = int(row['score_qos'] >= qos_seuil and tv <= 0.80 and td <= 0.80)
        for feat in self.features:
            if feat not in row: row[feat] = 0.0
        return row

    def _action_rl(self, row: dict) -> str:
        if self.agent_rl is None or self.env_rl is None: return 'N/A'
        eps_save = self.agent_rl.epsilon
        self.agent_rl.epsilon = 0.0
        state  = self.env_rl._get_state(row)
        sq     = float(row.get('score_qos', 1.0))
        qok    = int(row.get('optimisation_qos_autorisee', 1))
        va     = get_valid_actions(sq, qok, QOS_MIN_LIST, self.env_rl.n_actions)
        action = self.agent_rl.greedy_action(state, va)
        self.agent_rl.epsilon = eps_save
        qm = self.env_rl.QOS_MIN[action]
        if qm is not None and (sq < qm or qok == 0):
            return 'aucune_action'
        return self.env_rl.ACTIONS[action]

    def predire(self, row_brut: dict) -> dict:
        t0  = time.time()
        row = self._encoder(row_brut.copy())
        row = self._enrichir(row)

        X          = np.array([[row.get(f, 0.0) for f in self.features]])
        conso_pred = float(self.modele_lgbm.predict(X)[0])
        q10        = float(self.quantiles['q10'].predict(X)[0])
        q90        = float(self.quantiles['q90'].predict(X)[0])

        X_anom = np.array([[row.get(f, 0.0) for f in self.anom_features]])
        try:
            X_anom_scaled = self.scaler.transform(X_anom)
            score_iso     = float(-self.iso_f.score_samples(X_anom_scaled)[0])
        except Exception:
            score_iso = 0.5
        score_norm = np.clip((score_iso - 0.3) / 0.4, 0, 1)
        row['anomalie_score_ensemble'] = score_norm
        row['nb_votes_anomalie']       = int(score_norm > 0.5) * 3

        conso_reel      = row_brut.get('consommation_kwh', conso_pred)
        hors_intervalle = not (q10 <= conso_reel <= q90)
        decision        = self.moteur.decider_mode(pd.Series(row))
        score_qos       = float(row.get('score_qos', 1.0))
        type_anomalie   = ('prioritaire' if hors_intervalle and score_qos < qos_seuil
                           else 'energetique_pure' if hors_intervalle else 'normale')

        return {
            'timestamp'                 : datetime.now().isoformat(),
            'station_id'                : row_brut.get('station_id', 'INCONNUE'),
            'conso_predite_kwh'         : round(conso_pred, 4),
            'intervalle_q10'            : round(q10, 4),
            'intervalle_q90'            : round(q90, 4),
            'hors_intervalle'           : hors_intervalle,
            'score_anomalie'            : round(score_norm, 4),
            'type_anomalie'             : type_anomalie,
            'mode_operation'            : decision['mode'],
            'action_proposee_regles'    : decision['action_principale'],
            'action_proposee_rl'        : self._action_rl(row),
            'agent_rl_utilise'          : best_name,
            'economie_kwh_estimee'      : round(decision['eco_potentiel_pct'] / 100 * conso_pred, 4),
            'score_qos'                 : round(score_qos, 4),
            'risque_qos'                : decision['risque_qos'],
            'optimisation_qos_autorisee': int(row.get('optimisation_qos_autorisee', 1)),
            'latence_ms'                : round((time.time()-t0)*1000, 2),
        }


pipeline = PipelineInferenceTempsReel(
    modele_lgbm=modele_lgbm, modele_anom=modele_anom, encodeurs=encodeurs,
    quantiles=quantiles, features=FEATURES, moteur_decision=moteur,
    agent_rl=best_agent, env_rl=env,
)

print(f'Pipeline d\'inférence instancié (agent RL : {best_name})')
print('\nSIMULATION — 5 INFÉRENCES TEMPS RÉEL\n')

exemples = [
    {'station_id': 'TT-TUN-012', 'heure': 3,  'mois': 7, 'jour_semaine': 1,
     'technologie': '4G+', 'type_zone': 'Urbain', 'charge_cpu_pct': 20,
     'trafic_data_mbps': 5, 'temperature_ambiante': 32, 'nb_utilisateurs_actifs': 10,
     'puissance_emission_dbm': 43, 'nb_secteurs_actifs': 3, 'trafic_voix_erlang': 2,
     'taux_charge_voix': 0.03, 'taux_charge_data': 0.03, 'score_qos': 0.87,
     'optimisation_qos_autorisee': 1, 'est_weekend': 0, 'est_ferie': 0,
     'consommation_kwh': 3.5, 'conso_moy_24h': 4.0, 'conso_moy_7j': 4.1, 'eei': 95},
    {'station_id': 'TT-GAB-005', 'heure': 14, 'mois': 7, 'jour_semaine': 4,
     'technologie': '4G',  'type_zone': 'Urbain', 'charge_cpu_pct': 88,
     'trafic_data_mbps': 95, 'temperature_ambiante': 38, 'nb_utilisateurs_actifs': 420,
     'puissance_emission_dbm': 46, 'nb_secteurs_actifs': 6, 'trafic_voix_erlang': 45,
     'taux_charge_voix': 0.56, 'taux_charge_data': 0.95, 'score_qos': 0.38,
     'optimisation_qos_autorisee': 0, 'est_weekend': 0, 'est_ferie': 0,
     'consommation_kwh': 12.8, 'conso_moy_24h': 8.5, 'conso_moy_7j': 8.0, 'eei': 165},
    {'station_id': 'TT-SFA-001', 'heure': 2,  'mois': 5, 'jour_semaine': 3,
     'technologie': '4G',  'type_zone': 'Periurbain', 'charge_cpu_pct': 15,
     'trafic_data_mbps': 3, 'temperature_ambiante': 12, 'nb_utilisateurs_actifs': 8,
     'puissance_emission_dbm': 42, 'nb_secteurs_actifs': 4, 'trafic_voix_erlang': 1,
     'taux_charge_voix': 0.01, 'taux_charge_data': 0.03, 'score_qos': 0.73,
     'optimisation_qos_autorisee': 1, 'est_weekend': 0, 'est_ferie': 0,
     'consommation_kwh': 3.2, 'conso_moy_24h': 3.5, 'conso_moy_7j': 3.6, 'eei': 88},
    {'station_id': 'TT-BIS-003', 'heure': 3,  'mois': 1, 'jour_semaine': 6,
     'technologie': '4G+', 'type_zone': 'Urbain', 'charge_cpu_pct': 18,
     'trafic_data_mbps': 2, 'temperature_ambiante': 11, 'nb_utilisateurs_actifs': 6,
     'puissance_emission_dbm': 41, 'nb_secteurs_actifs': 3, 'trafic_voix_erlang': 1,
     'taux_charge_voix': 0.01, 'taux_charge_data': 0.01, 'score_qos': 0.91,
     'optimisation_qos_autorisee': 1, 'est_weekend': 1, 'est_ferie': 0,
     'consommation_kwh': 4.1, 'conso_moy_24h': 4.3, 'conso_moy_7j': 4.2, 'eei': 89,
     'vitesse_vent_ms': 5.0, 'humidite_relative_pct': 60, 'efficacite_free_cooling': 0.65},
    {'station_id': 'TT-KAI-009', 'heure': 11, 'mois': 8, 'jour_semaine': 2,
     'technologie': '2G',  'type_zone': 'Rural', 'charge_cpu_pct': 72,
     'trafic_data_mbps': 0.2, 'temperature_ambiante': 38, 'nb_utilisateurs_actifs': 110,
     'puissance_emission_dbm': 43, 'nb_secteurs_actifs': 2, 'trafic_voix_erlang': 28,
     'taux_charge_voix': 0.93, 'taux_charge_data': 0.40, 'score_qos': 0.35,
     'optimisation_qos_autorisee': 0, 'est_weekend': 0, 'est_ferie': 0,
     'consommation_kwh': 2.1, 'conso_moy_24h': 2.0, 'conso_moy_7j': 2.0, 'eei': 140},
]

print(f'{"Station":<14} {"QoS":>6} {"Risque":>8} {"Mode":<10} {"Action Règles":<28} {"Action RL":<28} {"Lat ms":>7}')
print('-' * 108)
resultats_sim = []
for ex in exemples:
    res = pipeline.predire(ex)
    resultats_sim.append(res)
    print(f"{res['station_id']:<14} "
          f"{res['score_qos']:>6.3f} "
          f"{res['risque_qos']:>8s} "
          f"{res['mode_operation']:<10} "
          f"{res['action_proposee_regles'][:26]:<28} "
          f"{res['action_proposee_rl'][:26]:<28} "
          f"{res['latence_ms']:>7.1f}")
print()
print(f'Latence moyenne : {np.mean([r["latence_ms"] for r in resultats_sim]):.1f} ms')
print(f'Agent RL utilisé : {best_name}')
```

## Limitations & perspectives

### Limitations de l'approche actuelle

| Dimension | Limitation | Impact |
|---|---|---|
| **RL tabulaire** | Espace d'état discret (288 états) — simplification des variables continues (QoS, charge) | Perte d'information sur les états intermédiaires |
| **Simulation** | L'environnement rejoue l'historique `df_train_rl` sans stochasticité réelle des transitions | La politique peut sur-adapter à la distribution historique |
| **Récompense** | Les fractions d'économie par action (ECO_FRACTION) sont fixes — non apprises | Sous-optimal pour les stations atypiques |
| **Dynamique QoS** | ±0.01/−0.05 par step est une approximation — non calibrée sur la dégradation réelle observée | Peut produire des décisions trop ou pas assez prudentes |
| **Évaluation** | L'évaluation statique ignore le carry QoS inter-steps | Légère sous-estimation des économies réelles |

### Perspectives d'amélioration

1. **Deep Q-Network (DQN)** : remplacer la Q-table par un réseau de neurones pour traiter
   les états continus — capturer des nuances non visibles avec 288 états discrets.

2. **Calibration de l'environnement** : ajuster les paramètres de dynamique QoS (±Δ par step)
   sur les transitions réellement observées dans `df_2024` (régression sur Δ_QoS ~ actions).

3. **Déploiement en ligne** : passer d'une simulation historique à un apprentissage en ligne
   avec retour de l'infrastructure réelle (API NOC Tunisie Telecom).

4. **Multi-agent** : modéliser chaque BTS comme un agent indépendant avec coordination locale
   pour éviter la sur-extinction simultanée de secteurs adjacents.
